# 02 — 单电子运动轨迹与动能

对应 Rust 主程序：`cargo run --release`。读取 `results/trajectories/electron.csv`，绘制 XY 投影、三维轨迹和动能变化。

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 从 Jupyter 当前工作目录向上搜索 Cargo.toml，定位项目根目录
current = Path.cwd().resolve()
PROJECT_ROOT = next((p for p in (current, *current.parents) if (p / 'Cargo.toml').exists()), None)
if PROJECT_ROOT is None:
    raise FileNotFoundError('未找到 Cargo.toml，请从 Rust 项目下启动 Jupyter')
print('项目根目录:', PROJECT_ROOT)
```

```python
csv_path = PROJECT_ROOT / 'results' / 'trajectories' / 'electron.csv'
if not csv_path.exists():
    raise FileNotFoundError(f'缺少 {csv_path}，先运行 cargo run --release')
df = pd.read_csv(csv_path)
display(df.head())
print('轨迹点数:', len(df))
```

```python
x, y, z = [df[k].to_numpy() * 1e3 for k in ('x_m','y_m','z_m')]
time_ns = df['time_s'].to_numpy() * 1e9
```

```python
fig, ax = plt.subplots(figsize=(7, 7))
ax.plot(x, y, lw=0.7)
ax.set(xlabel='X (mm)', ylabel='Y (mm)', title='Electron orbit — XY projection')
ax.set_aspect('equal', adjustable='box')
ax.grid(True, alpha=0.3)
fig.tight_layout()
plt.show()
```

```python
fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection='3d')
ax.plot(x, y, z, lw=0.7)
ax.set(xlabel='X (mm)', ylabel='Y (mm)', zlabel='Z (mm)', title='Electron trajectory')
fig.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(time_ns, df['energy_eV'], lw=1.2)
ax.set(xlabel='Time (ns)', ylabel='Kinetic energy (eV)', title='Electron kinetic energy')
ax.grid(True, alpha=0.3)
fig.tight_layout()
plt.show()
```

**物理解释：** 纯磁场中洛伦兹力不做功，因此动能应守恒；本例包含 13.56 MHz 射频电场，电子动能可能随时间变化。
