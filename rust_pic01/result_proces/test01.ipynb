```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/Users/shyee/code/rust_learning/rust_pic01/results/rf_field_test.csv")

plt.figure(figsize=(10, 5))

plt.plot(
    df["time_ns"],
    df["Ex_V_m"],
    label="RF Electric Field"
)

plt.xlabel("Time (ns)")
plt.ylabel("Ex (V/m)")

plt.title("13.56 MHz RF Electric Field")

plt.grid(True)
plt.legend()

plt.show()
```
